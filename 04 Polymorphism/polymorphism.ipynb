#### Polymorphism

    Same operator can be worked on different classes
    Same method can be applied on different classes

```python
a = [1, 2, 3]
print(type(a))
print(len(a))
a = [1, 2, 3, "d", "d"]
print(type(a))
print(len(a))
```

```python
a = {1: 1, 2: 2}
print(type(a))
print(len(a))
```

```python
a = (1, 1, 2, 2)
print(type(a))
print(len(a))
```

```python
a = [1, 2] + ["a"] + [1.0]
print(type(a))
print(len(a))
```

```python
a = "a" + "1.0"
print(type(a))
print(len(a))
```

```python
class India:
    def capital(self):
        print("New Delhi is capital of India")

    def language(self):
        print("Hindi is widely spoken in India")


class USA:
    def capital(self):
        print("Washington D.C is capital of USA")

    def language(self):
        print("English is widely spoken in USA")


class France:
    def capital(self):
        print("Paris is capital of France")

    def language(self):
        print("French is widely spoken in France")
```

```python
c1 = India()
type(c1)
c1.capital()
c1.language()
```

```python
c2 = USA()
print(type(c2))
c2.capital()
c2.language()
```

```python
country = [India(), USA(), France()]
for c in country:
    c.capital()
    c.language()
    print("\n=====")
```

```python
from dataclasses import dataclass
import math


@dataclass
class Circle:
    radius: int | float

    def perimeter(self):
        return 2 * math.pi * self.radius

    def area(self):
        return math.pi * (self.radius**2)


@dataclass
class Rectangle:
    length: int | float
    breadth: int | float

    def perimeter(self):
        return 2 * (self.length + self.breadth)

    def area(self):
        return self.length * self.breadth


@dataclass
class Triangle:
    a: int | float
    b: int | float
    c: int | float

    def perimeter(self):
        return self.a + self.b + self.c

    def area(self):
        semi_perimeter = self.perimeter() / 2

        area = math.sqrt(
            semi_perimeter
            * (semi_perimeter - self.a)
            * (semi_perimeter - self.b)
            * (semi_perimeter - self.c)
        )

        return area
```

```python
shapes = [Rectangle(1, 2), Circle(2.34), Triangle(7, 10, 6)]
for shape in shapes:
    print(shape)
print("\n=======")
for shape in shapes:
    print(shape)
    print(f"Perimeter: {shape.perimeter()}, Area: {shape.area()}")
    print("\n*******")
```

```python
import numpy as np


@dataclass
class NumericalSummary:

    def __init__(self, data):
        self.data = data

    def summary(self):
        return {
            "mean": np.mean(self.data),
            "max": np.max(self.data),
            "min": np.min(self.data),
            "std": np.std(self.data),
        }


@dataclass
class CategoricalSummary:

    def __init__(self, data):
        self.data = data

    def summary(self):
        unique_values, count = np.unique(self.data, return_counts=True)
        return dict(zip(unique_values, count))
```

```python
numericals = [1, 2, 3]
numerical_summary = NumericalSummary(numericals)
print(numerical_summary)
print(numerical_summary.summary())
print("\n=====")

categoricals = ["1", "2", "3"]
categorical_summary = CategoricalSummary(categoricals)
print(categorical_summary)
print(categorical_summary.summary())
print("\n===")
```

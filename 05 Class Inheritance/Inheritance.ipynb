#### Inheritane

    1. Single Inheritance
        One parent one child

```python
class Person:
    def __init__(self, name: str, age: int):
        self.name = name
        self.age = age

    def intro(self):
        print(f"My name is {self.name}")
        print(f"I am {self.age} old")


class Employee(Person):
    def __init__(self, name: str, age: int, company: str):
        super().__init__(name, age)
        self.company = company

    def work(self):
        print(f"I work for {self.company}")


employee = Employee(name="Deanesh", age=46, company="Startup")
employee.intro()
employee.work()
```

##### Conditons in pydantic

    ge - greater than or equal to
    le - less than or equal to
    gt - greater than
    lt - less than
    max_length - maximum length

```python
# Base model in pydantic is responsible for validating data
# Field is used to add some checks to data
from pydantic import BaseModel, Field
from typing import Literal


class Student(BaseModel):
    roll_no: int = Field(ge=101, le=1000)
    name: str = Field(max_length=20)
    age: int = Field(ge=18, le=60)
    marks: float = Field(ge=0.0, le=100.0)
    gender: Literal["male", "female"]
    hobbies: list[str] = Field(default_factory=list)

    def get_student_info(self):
        print(f"Roll Number : {self.roll_no}")
        print(f"Name : {self.name}")
        print(f"Marks : {self.marks}")
        print(f"Gender : {self.gender}")
        print(f"Hobbies : {self.hobbies}")
```

```python
s1 = Student(
    roll_no=101,
    name="Raman",
    age=25,
    marks=85.3,
    gender="male",
    hobbies=["cricket", "singing"],
)
```

```python
s1.roll_no, s1.gender, s1.age, s1.hobbies
```

```python
s1.get_student_info()
```

```python
# Base model in pydantic is responsible for validating data
# Field is used to add some checks to data
from pydantic import BaseModel, Field
from typing import Literal


class AgeValidator(BaseModel):
    name: str = Field(max_length=20)
    age: int = Field(ge=18, le=55)

    def get_validated_age(self):
        print(f"Validated Age: {self.age}")


age = AgeValidator(name="123", age=20)
age
age.model_dump()
```

#### Mutiple Inheritance


   (Base Salary, Bonus, Tax) -> (Total Salary)



```python
class BaseSalary:

    def __init__(self, salary: float):
        self.salary = salary

    def calculate_base_salary(self, salary: float):
        return salary


class Bonus:

    def __init__(self, bonus_perc: float):
        self.bonus_perc = bonus_perc

    def calculate_bonus(self, salary: float):
        return (self.bonus_perc / 100) * salary


class Tax:

    def __init__(self, tax_perc: float):
        self.tax_perc = tax_perc

    def calculate_tax(self, salary: float):
        return (self.tax_perc / 100) * salary


class TotalTaxSalary(BaseSalary, Bonus, Tax):
    def __init__(self, salary, bonus_perc, tax_perc):
        BaseSalary.__init__(self, salary)
        Bonus.__init__(self, bonus_perc)
        Tax.__init__(self, tax_perc)

    def calculate_total_salary(self):
        salary = self.calculate_base_salary(self.salary)
        bonus_amount = self.calculate_bonus(self.salary)
        tax_amount = self.calculate_tax(self.salary)
        gross_salary = self.salary + bonus_amount
        net_sal = gross_salary - tax_amount

        return {
            "Base Salary": salary,
            "Bonus": bonus_amount,
            "Gross Salary": gross_salary,
            "Tax": tax_amount,
            "Net Salary": net_sal,
        }


emp_sal_list = [
    TotalTaxSalary(10000, 12, 10),
    TotalTaxSalary(678000, 14.5, 11.25),
    TotalTaxSalary(988000, 14.5, 11.25),
]
for emp_sal in emp_sal_list:
    print(emp_sal.calculate_total_salary())
```

#### Multilevel Inheritance
    Employee (emp_id, name) -> Manager (emp_id, name, dept_id) -> Project Manager (emp_id, name, dept_id, project_id)

```python
class Employee:
    def __init__(self, emp_id: int, name: str):
        self.emp_id = emp_id
        self.name = name

    def employee_details(self):
        return {"Employee ID": self.emp_id, "Name": self.name}


class Manager(Employee):
    def __init__(self, emp_id: int, name: str, dept_id: int):
        super().__init__(emp_id, name)
        self.dept_id = dept_id

    def manager_details(self):
        manager_dict = self.employee_details()
        manager_dict["Department ID"] = self.dept_id
        return manager_dict


class ProjectManager(Manager):
    def __init__(self, emp_id: int, name: str, dept_id: int, project_id: int):
        super().__init__(emp_id, name, dept_id)
        self.project_id = project_id

    def project_manager_details(self):
        project_manager_dict = self.manager_details()
        project_manager_dict["Project Manager ID"] = self.project_id
        return project_manager_dict
```

```python
project_manager_list = [
    ProjectManager(123, "DT", 1, 2),
    ProjectManager(1, "RT", 2, 2),
    ProjectManager(27, "DHT", 3, 1),
]

print("Project Manager Details .. Start")
for project_manager in project_manager_list:
    print("\n")
    for key, val in project_manager.project_manager_details().items():
        print(f"{key}:{val}")

print("\nProject Manager Details .. End")
```

#### Hierarchial Inheritance
    Shape(area, perimeter, info(print area , perimeter)) 
        -> Rectangle(len, wid, area, perimeter)
        -> Square(side, area, perimeter)


```python
class Shape:
    def area(self):
        pass

    def perimeter(self):
        pass

    def info(self):
        return {
            "Class": type(self).__name__,
            "Area": self.area(),
            "Perimeter": self.perimeter(),
        }


class Rectangle(Shape):
    def __init__(self, length: float, width: float):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)


class Square(Shape):
    def __init__(self, side: float):
        self.side = side

    def area(self):
        return self.side**2

    def perimeter(self):
        return 4 * self.side


shapes = [Rectangle(10, 20), Rectangle(11, 10), Square(3), Square(5)]

for shape in shapes:
    shapes_info = {}
    shapes_info["Shape_Size_Info"] = shape.info()
    print(shapes_info)
```

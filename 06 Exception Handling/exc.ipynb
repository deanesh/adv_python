#### Method without exception handling

#### Method with exception handling

```python
def division2(a: int | float, b: int | float) -> float | None:
    try:
        c = a / b
        return c
    except Exception as ex:
        print(f"Exception occured: {ex}")


d1 = division2(23, 45)
print(d1)
print("completed")
```

```python
d2 = division2(23, 0)
print(d2)
print("completed")
```

```python
d3 = division2("23", "0")
print(d2)
print("completed")
```

#### Use different exception for each errors

```python
import math


def division3(a: int | float, b: int | float) -> float | None:
    try:
        c = a / b
        return c
    except ZeroDivisionError as zde:
        print(f"Denominator cannot be zero: {zde}")
    except TypeError as te:
        print(
            f"Invalid inputs . Provide numbers as input and denominator should not be zero: {te}"
        )
    except Exception as ex:
        print(f"Unexpected exception occured: {ex}")
    finally:
        print("Method Completed")
```

```python
def age_validator(age: int | float) -> float | None:
    try:
        if age <= 0 or age >= 100:
            raise ValueError("Age should be between 0 and 100")
        return age
    except Exception as ex:
        print(f"Invalid age occured: {ex}")
    finally:
        print("Method Completed")
```

```python
age_2 = age_validator(-70)
if age_2:
    print(age_2)
print("completed")
```

```python
try:
    input_age = int(input("Please enter age: "))
    validated_age = age_validator(input_age)
    print(f"Valid age entered: {validated_age}")
except Exception as ex:
    print(f"Invalid age entered: {ex}")
```

```python
import math


def square_root(num: int | float) -> float | None:
    try:
        s = math.sqrt(num)
    except Exception as ex:
        print(f"Exception occurred: {ex}")
    else:
        print(f"Square root of number is {s}")
        return s
    finally:
        print("Method completed")
```

```python
res = square_root(24)
```

```python
res1 = square_root(-23)
```

```python
res2 = square_root([-23])
```

#### File Handling

```python
# Append to file
with open("new.txt", "a", encoding="utf-8") as f:
    f.write("\nI am learning ML/AI Topics")
    f.write("\nTheory is understood. Practice needs to be done")
```

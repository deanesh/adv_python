#### Numpy Arrays

    1D Array
    2D Array
    3D Array

```python
import numpy as np

# 1D Array
a = [1, 2, 3]
print(type(a))
a1 = np.array(a)
print(type(a1))
print(a1.shape, a1.size, a1.dtype)
```

```python
import numpy as np

# 2D Array
a_2d = [[1, 2, 3, 4], [4, 5, 6, 4], [7, 8, 9, 4]]
print(type(a_2d))
np_2d = np.array(a_2d)
print(type((np_2d)))
print(np_2d.shape, np_2d.dtype, np_2d.size)
```

#### 3D Array

```python
a_3d = [[[1, 2, 3], [4, 5, 6]], [[7, 8, 9], [0, 1, 2]]]
print(type(a_3d))
np_3d = np.array(a_3d)
print(type(np_3d))
print(np_3d.shape, np_3d.size, np_3d.dtype)

class_a_st_ranks = ["a", "b"]
class_a_st_2_ranks = ["c", "d"]
class_b_st_3_ranks = ["e", "f"]
class_b_st_4_ranks = ["g", "h"]
class_a = [class_a_st_ranks, class_a_st_2_ranks]
class_b = [class_b_st_3_ranks, class_b_st_4_ranks]
a_3d_str = [class_a, class_b]
print(type(a_3d_str))
np_3d_str = np.array(a_3d_str)
print(type(np_3d_str))
print(np_3d_str.shape, np_3d_str.size, np_3d_str.dtype)
```

#### List vs Numpy

```python
# On NP Array looping is not required for multiplication
a = [1, 2, 3]
a1 = a * 3
print(a1)
np_a = np.array(a)
np_a1 = np_a * 3
print(np_a1)
```

```python
# On NP Array looping is not required for addition
# list will not support
try:
    np_a = np.array(a)
    np_a1 = np_a + 3
    print(np_a1)
    a = [1, 2, 3]
    a1 = a + 3
    print(a1)
except Exception as ex:
    print(ex)
```

#### Aggregate Functions on Numpy
    Sum
    Mean
    Mode
    Median
    Count
    Std

```python
a = np.array(range(1, 30, 6))
sum_ = np.sum(a)
mean_ = np.mean(a)
max_ = np.max(a)
min_ = np.min(a)
std_ = np.std(a)
count_ = np.size(a)
print(min_, max_, count_, mean_, std_)
```

```python
a_3d = [[[1, 2, 3], [4, 5, 6]], [[7, 8, 9], [0, 1, 2]]]
np_3d = np.array(a_3d)

axis_0_sum = np_3d.sum(axis=0)
axis_0_sum
axis_1_sum = np_3d.sum(axis=1)
axis_0_sum, axis_1_sum

axis_0_prod = np_3d.prod(axis=0)
axis_0_prod
axis_1_prod = np_3d.prod(axis=1)
axis_0_prod, axis_1_prod
```

#### Calculate sin, cos , tan you require to conver tot radian

```python
# in order to calculate sin, cos , tan you require to conver tot radian
angle_deg = [0, 30, 45, 60, 90]
print(angle_deg)
angle_rad = np.radians(angle_deg)
print(angle_rad.round(3))
```

```python
np.sin(angle_rad).round(3)
```

```python
np.cos(angle_rad).round(3)
```

```python
np.tan(angle_rad.round(3))
```

```python
x = [0.1, 0.3, 1, 0.5, 0.8]
rad = np.asin(x)
print(rad.round(3))
deg = np.degrees(rad)
print(deg.round(3))
```

```python
np.acos(x), np.atan(x)
```

```python
import types
import inspect


# --------------------------------------------------
# 📘 List all attribute names of an object/module/class
def list_all(obj):
    """
    Example:
        import sklearn
        list_all(sklearn)
    """
    return dir(obj)


# --------------------------------------------------
# 🔧 List only functions (user-defined or built-in)
def list_functions(obj):
    """
    Example:
        import sklearn.feature_selection as fs
        list_functions(fs)
    """
    return [
        name
        for name in dir(obj)
        if isinstance(
            getattr(obj, name), (types.FunctionType, types.BuiltinFunctionType)
        )
    ]


# --------------------------------------------------
# 🏷️ List only classes
def list_classes(obj):
    """
    Example:
        import sklearn.feature_selection as fs
        list_classes(fs)
    """
    return [name for name in dir(obj) if isinstance(getattr(obj, name), type)]


# --------------------------------------------------
# 📦 List only submodules
def list_submodules(obj):
    """
    Example:
        import sklearn
        list_submodules(sklearn)
    """
    return [
        name for name in dir(obj) if isinstance(getattr(obj, name), types.ModuleType)
    ]


# --------------------------------------------------
# 🧱 List only variables/constants (not funcs, classes, or modules)
def list_variables(obj):
    """
    Example:
        import sklearn.feature_selection as fs
        list_variables(fs)
    """
    return [
        name
        for name in dir(obj)
        if not name.startswith("__")
        and not isinstance(
            getattr(obj, name),
            (types.FunctionType, types.BuiltinFunctionType, types.ModuleType, type),
        )
    ]


# --------------------------------------------------
# 🖋️ Get the signature of a function/method
def get_signature(obj, name):
    """
    Example:
        import sklearn.feature_selection as fs
        get_signature(fs, 'mutual_info_classif')
    """
    try:
        return str(inspect.signature(getattr(obj, name)))
    except Exception as e:
        return f"Cannot get signature: {e}"


# --------------------------------------------------
# 📄 Get the docstring of any attribute
def get_doc(obj, name):
    """
    Example:
        import sklearn.feature_selection as fs
        get_doc(fs, 'mutual_info_classif')
    """
    try:
        return inspect.getdoc(getattr(obj, name))
    except Exception as e:
        return f"Cannot get doc: {e}"
```

```python
import numpy as np

entity = np.ndarray

print("🔧 Functions:", list_functions(entity))
print("🏷️ Classes:", list_classes(entity))
print("📦 Submodules:", list_submodules(entity))
print("🧱 Variables:", list_variables(entity))
```

```python
entity = "ndarray"
pkg = np
print(f"\n🖋️ Signature of {entity}:")
print(get_signature(pkg, entity))

print(f"\n📄 Docstring of {entity}:")
print(get_doc(pkg, entity))
```

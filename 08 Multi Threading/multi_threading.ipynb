#### Multi Threading
    Single Threaded - One function can be executed at a time
    Multi Threaded - Multiple functions can be executed at a time

```python
import time
```

```python
%%time
time.sleep(2)
print("Hello")
time.sleep(3)
print("World")
```

```python
def simple_interest(p, n, r):
    print("Simple Interest function started...")
    time.sleep(2)
    i = (p * n * r) / 100
    a = p + i
    print(f"Simple Interest: {i:.2f} INR")
    print("Simple Interest function completed...")

```

```python
%%time 
simple_interest(p=10000, n=5, r=6)
```

```python
import math


def hypotenuse(a, b):
    print("Hypotenuse function started...")
    time.sleep(3)
    c = math.sqrt(a**2 + b**2)
    print(f"Hypotenuse of {a} and {b} is {c:.2f}")
```

```python

%%time
hypotenuse(3, 4)
```

```python
%%time 
simple_interest(p=10000, n=5, r=6)
hypotenuse(3, 4)
```

##### Multi Threading - To execute functions simultaneously

```python
%%time
from threading import Thread

# Create a thread for simple interest
th1 = Thread(target=simple_interest, args=(100000, 7, 5.25))
th2 = Thread(target=hypotenuse, args=(3, 4))

# start threads
th1.start()
th2.start()


# wait for threads to complete
th1.join()
th2.join()
```

```python
def average(num_1, num_2):
    print("avg start")
    time.sleep(1)
    avg = (num_1 + num_2) / 2
    print(f"avg = {avg:.2f}")


def multiply(num_1, num_2):
    print("multiply of 2 numbers start")
    time.sleep(1.305)
    mul = num_1 * num_2
    print(f"multiply of 2 numbers: {mul:.2f}")
```

```python
%%time
average(12, 34)
multiply(10, 20)
```

```python
%%time
from threading import Thread

# Instantiate threads
t1 = Thread(target=average, args=(107, 290.3456))
t2 = Thread(target=multiply, args=(10.2, 20.2))

# start threads
t1.start()
t2.start()

# wait for threads to complete
t1.join()
t2.join()
```

##### Same function to calculate multiple values parallely

```python
def square(n):
    print("Square fucntion start...")
    time.sleep(2)
    s = n**2
    print(f"Square fucntion result: {s:.2f}")
```

```python
numbers = [2, 3, 4, 5, 6, 7, 8, 9]
len(numbers)
```

```python
def square_of_list(a: list) -> list:
    for i in a:
        square(i)
```

```python
%%time
square_of_list(numbers)
```

```python
import time
from threading import Thread


def square(n):
    print("Square fucntion start...")
    time.sleep(2)
    s = n**2
    print(f"Square fucntion result: {s:.2f}")


def square_multithreading(a: list):
    print("square multi threading start...")
    # Create a blank list
    threads = []
    for i in a:
        th = Thread(target=square, args=(i,))
        threads.append(th)

    # wait for threads to start
    for th in threads:
        th.start()

    # wait for threads to stop
    for th in threads:
        th.join()

    print("square multi threading end...")


```

```python
%%time
square_multithreading(numbers)
```

```python
import time
from threading import Thread
from urllib.request import urlretrieve


def download_file(url, filename):
    print(f"Download started: {filename}")
    time.sleep(1)  # Simulate delay (optional)
    urlretrieve(url, filename)
    print(f"Download finished: {filename}")


def download_multithreading(urls: list):
    print("Multithreaded downloading start...")
    threads = []

    for i, url in enumerate(urls):
        filename = f"file_{i + 1}.html"  # Save as HTML for demonstration
        th = Thread(target=download_file, args=(url, filename))
        threads.append(th)

    for th in threads:
        th.start()

    for th in threads:
        th.join()

    print("Multithreaded downloading end...")
```

```python
import time
from threading import Thread
from urllib.request import urlretrieve


def download_multithreaded(urls: list[str]):
    threads = []

    for url in urls:
        th = Thread(target=download_file, args=(url,))
        th.start()
        threads.append(th)

    for th in threads:
        th.join()
```

#### Creating a class for wikpedia scrapper

```python
import requests
from bs4 import BeautifulSoup
import IPython.display
from io import StringIO
import pandas as pd
import lxml


class WikiScraper:

    def __init__(self, url: str):
        self.url = url
        header = {"User-Agent": "Chrome/127.0.0.1"}
        response = requests.get(self.url, headers=header)
        response.raise_for_status()
        self.soup = BeautifulSoup(response.content)

    def get_title(self):
        title_tag = self.soup.find(name="title")
        return title_tag.text

    def get_heading(self):
        heading_tag = self.soup.find(name="h1", class_="firstHeading")
        return heading_tag.text

    def get_subheading(self):
        sub_tags = self.soup.find_all("div", class_="mw-heading")
        sub_text = [tag.text for tag in sub_tags]
        return sub_text

    def get_paras(self):
        p_tags = self.soup.find_all(name="p")
        p_text = [tag.text for tag in p_tags]
        return p_text

    def save_text(self, text: list[str], filename: str):
        with open(filename, "w", encoding="utf-8") as f:
            f.writelines(text)

    def get_image_links(self):
        a_tags = self.soup.find_all(name="a", class_="mw-file-description")
        home_page = "https://en.wikpedia.org"
        img_urls = [home_page + tag.get("href") for tag in a_tags]
        return img_urls

    def get_tables(self):
        table_tags = self.soup.find_all(name="table", class_="wikitable")
        dfs = []
        for table in table_tags:
            t = StringIO(str(table))
            df = pd.read_html(t)[0]
            dfs.append(df)
        return dfs
```

#### Get image links from different wikipedia urls

```python
urls = [
    "https://en.wikipedia.org/wiki/Data_science",
    "https://en.wikipedia.org/wiki/Data_analysis",
    "https://en.wikipedia.org/wiki/World_population",
    "https://en.wikipedia.org/wiki/Python_(programming_language)",
]
```

```python
img_dict = {}

for url in urls:
    print(f"Scraping : {url}")
    scraper = WikiScraper(url)
    h1_text = scraper.get_heading()
    img_links = scraper.get_image_links()
    img_dict[h1_text] = img_links

img_dict
```

```python
import json

with open("images.json", "w") as f:
    content = json.dumps(img_dict, indent=4)
    f.write(content)
```

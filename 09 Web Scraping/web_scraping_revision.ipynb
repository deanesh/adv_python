```python
import pandas as pd
import IPython.display
import requests
from bs4 import BeautifulSoup


header = {"User-Agent": "Chrome/127.0.0.1"}
url = "https://en.wikipedia.org/wiki/World_population"
response = requests.get(url=url, headers=header)
response.raise_for_status
soup = BeautifulSoup(response.content)
```

```python
soup.contents[0]
```

```python
# Get Title
title_tag = soup.find(name="title")
title_tag.text
```

```python
# Get Heading
heading_tag = soup.find(name="h1", class_="firstHeading")
heading_tag.text
```

```python
# Get Sub Headings
subheadings = soup.find_all(name="div", class_="mw-heading")
subheadings_list = [tag.text for tag in subheadings]
subheadings_list.__len__()
```

```python
# Get Paragraphs
paragraphs = soup.find_all(name="p")
paragraphs_list = [tag.text for tag in paragraphs]
paragraphs_list.__len__()
```

```python
# Write paragraphs to text file
with open("paragraphs.txt", "w", encoding="utf-8") as f:
    f.writelines(paragraphs_list)
```

```python
# Get Images
images = soup.find_all(name="a", class_="mw-file-description")
home_page = "https://en.wikipedia.org"
image_urls_list = [home_page + tag.get("href") for tag in images]
image_urls_list.__len__()
```

```python
# Get Tables
from io import StringIO

tables = soup.find_all(name="table", class_="wikitable")
dfs = []
for table in tables:
    t = StringIO(str(table))
    df = pd.read_html(t)[0]
    dfs.append(df)

dfs[0].to_csv("table.csv", index=False)

for df in dfs:
    print(df)
    break
```
